Hierarchical Columns

```python
columns = pd.MultiIndex.from_tuples([
    ("Marks", "Midterm"),
    ("Marks", "Final"),
    ("Attendence", "Percentage")
])
score = pd.DataFrame(
    [[25, 62, 92], [22, 55, 65]],
    index = ["Aman", "Riya"],
    columns = columns
)
print(score)
```

```python
class_a = pd.DataFrame({
    "Student_Id":[1, 2],
    "Name": ["Aman", "Riya"],
    "Marks": [76, 88]
})
class_b = pd.DataFrame({
    "Student_Id":[3, 4],
    "Name": ["Kabir", "Neha"],
    "Marks": [82, 93]
})
```

```python
all_students = pd.concat([class_a, class_b], ignore_index = True)
print(all_students)
```

```python
result = pd.concat([class_a, class_b], axis = 1)
print(result)
```

```python
student_info = pd.DataFrame({
    "Student_Id":[1, 2, 3, 4],
    "Name": ["Aman", "Riya", "Kabir", "Neha"],
    "Age": [17, 16, 18, 17]
})
student_marks = pd.DataFrame({
    "Student_Id":[1, 2, 3, 4],
    "Subject": ["Math", "Science", "English", "Hindi"],
    "Marks": [76, 88, 82, 93]
})
```

```python
merged = pd.merge(student_info, student_marks, on = "Student_Id", how = "inner")
print(merged)
merged = pd.merge(student_info, student_marks, on = "Student_Id", how = "left")
print(merged)
merged = pd.merge(student_info, student_marks, on = "Student_Id", how = "right")
print(merged)
merged = pd.merge(student_info, student_marks, on = "Student_Id", how = "outer")
print(merged)
```

Merging when key-column names differ

```python
student_info = pd.DataFrame({
    "Student_Id":[1, 2, 3, 4],
    "Name": ["Aman", "Riya", "Kabir", "Neha"],
    "Age": [17, 16, 18, 17]
})
student_marks = pd.DataFrame({
    "ID":[1, 2, 3, 4],
    "Subject": ["Math", "Science", "English", "Hindi"],
    "Marks": [76, 88, 82, 93]
})
ans = pd.merge(student_info, student_marks, left_on = "Student_Id", right_on = "ID", how = "inner")
print(ans)
```

Joining by index

```python
left = pd.DataFrame({
    "Dept": ["ECE", "CSE", "ME"]},
    index=[101, 102, 103]
)
right = pd.DataFrame({
    "Marks": [82, 93, 76]},
    index=[85, 87, 91]
)
result = left.join(right)
print(result)
```

Reshaping and Pivoting

```python
wide = pd.DataFrame({
    "Student": ["Aman", "Riya", "Kabir"],
    "Math": [76, 88, 85],
    "Science": [82, 93, 91]
})
print(wide)
```

Convert wide data to Long using melt()

```python
long = wide.melt(id_vars = "Student",
                 var_name = "Subject",
                 value_name = "Marks")
print(long)
```

pivot_table()

```python
sales = pd.DataFrame({
    "Date": ["2023-01-01", "2023-01-01", "2023-01-02", "2023-01-02"],
    "Store": ["A", "B", "A", "B"],
    "Product": ["X", "Y", "X", "Y"],
    "Sales": [100, 200, 150, 300]
})
pivot_sales = sales.pivot_table(index = "Date",
                                columns = ["Store", "Product"],
                                values = "Sales",
                                aggfunc = "sum")
print(pivot_sales)
```

Cross Tabulation

```python
data = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", "Female"],
    "Age": [25, 30, 22, 28]
})
table = pd.crosstab(data["Gender"], data["Age"])
print(table)
```
